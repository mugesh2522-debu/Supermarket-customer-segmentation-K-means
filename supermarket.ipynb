```python
data['sales_revenue'].fillna(data['sales_revenue'].median(), inplace=True)

```

```python
data['units_sold'].fillna(data['units_sold'].median(), inplace=True)
```

```python
x = data[['customer_age', 'units_sold', 'sales_revenue']]
```

```python
from sklearn.cluster import KMeans
import warnings
warnings.filterwarnings("ignore")
wcss=[]
for i in range (1,11):
  kmeans=KMeans(n_clusters=i,random_state=40)
  kmeans.fit(x)
  wcss.append(kmeans.inertia_)
plt.plot(range(1,11),wcss)
plt.title('the Elbow method ')
plt.xlabel('Number of clusters')
plt.ylabel('WCSS')
plt.show()
```

```python
from sklearn.cluster import KMeans
model =KMeans(n_clusters=3,random_state=40)
model.fit(x)
```

```python
label=model.labels_
label
```
